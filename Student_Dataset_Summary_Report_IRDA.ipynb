## Dataset Summary Report- Day(7)

This report will explore the basic structure and characteristics of the `Student_Dataset.csv`.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/Student_Dataset.csv')

print('Dataset loaded successfully.')
```

### 1. Inspecting the First Few Rows

```python
display(df.head())
```

### 2. Inspecting the Last Few Rows

```python
display(df.tail())
```

### 3. Viewing Random Samples

To ensure a diverse understanding, let's display 5 random samples from the dataset.

```python
display(df.sample(n=5))
```

### 4. Dataset Shape (Rows and Columns)

The shape attribute tells us the number of rows and columns in the DataFrame.

```python
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
```

### 5. Column Names

These are the features available in the dataset.

```python
print("Column names:")
display(df.columns)
```

### 6. Index Information

The index provides labels for the rows.

```python
print("Index information:")
display(df.index)
```

### 7. Data Types of Columns

Understanding data types is crucial for data cleaning and analysis.

```python
print("Data types of columns:")
display(df.dtypes)
```

The `info()` method provides a concise summary of the DataFrame, including the number of non-null values and memory usage.

```python
df.info()
```

### 9. Descriptive Statistics with `describe()`

For numerical columns, `describe()` generates descriptive statistics such as count, mean, standard deviation, min, max, and quartiles.

```python
display(df.describe())
```
